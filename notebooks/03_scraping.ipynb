```python
from bs4 import BeautifulSoup
```

```python
with open("../data/raw/sample.html", encoding="utf-8") as f:
    html = f.read()
print(html)
```

```python
soup = BeautifulSoup(html, "html.parser")
```

```python
products = soup.find_all("div", class_="product")
```

```python
for p in products:
    name = p.find("h2", class_="name").text
    price = int(p.find("span", class_="price").text)
    print(name, price)
```

```python
for p in products:
    name = p.find("h2", class_="name").text
    price = int(p.find("span", class_="price").text)
    if price < 2000:
        print(name, price)
```

```python
records = []
for p in products:
    name = p.find("h2", class_="name").text
    price = int(p.find("span", class_="price").text)
    records.append({"name": name, "price": price})
```

```python
import pandas as pd 
df= pd.DataFrame(records)
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
df.shape
```

```python
df.to_csv("../data/raw/scraped_sample.csv", index=False)
```

```python
import requests
r = requests.get("https://books.toscrape.com/catalogue/category/books/travel_2/index.html")
r.encoding = "utf-8"
r.status_code
```

```python
soup = BeautifulSoup(r.text, "html.parser")
books = soup.find_all("article", class_="product_pod")
```

```python
records = []
for b in books:
    title = b.find("h3").find("a")["title"]
    price = float(b.find("p", class_="price_color").text.replace("£", ""))
    records.append({"title": title, "price": price})
len(records)
```

```python
df =pd.DataFrame(records)
```

```python
df.to_csv("../data/raw/books_travel.csv", index=False)
```
